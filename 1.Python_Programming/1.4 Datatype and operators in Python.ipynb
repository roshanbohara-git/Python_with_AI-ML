# Datatypes
- Data types are a classification of data which tell the compiler or interpreter how the programmer intends to use the data
- They determine the type of opertations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data

## Importance:
- Data types ensure that data is stored in an efficient way.
- They help in performing correct operations on data.
- Proper use of data types can prevent errors and bugs in the program.


```python
# Basic Data Types
# Interger Example

age = 35
type(age)
```

```python
# Float point data type
height = 5.8
print(height)
print(type(height))
```

```python
# String data type example
name = "Roshan"
print(name)
print(type(name))
```

```python
## boolean datatype
is_student = True
print(is_student)
print(type(is_student))
```

```python
a = 10
b = 10
type(a==b)
```

```python
## common errors we get while working with different daatypes 
result = "Hello"+5
```

```python
result = "Hello"+str(5)
print(result)
```

# Operators in Python

```python
## Arithmetic Operations
a = 10
b = 15
add_result = a+b # addition operation
sub_result = a-b # substraction operation
mult_result = a*b # multiplication operation
div_result = a/b # division operation
floor_div_result = a//b # floor division
modulus_result = a%b # modulus operation

exponent_result = a**b # Exponentiation
print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponent_result)

```

## Comparision Operators

```python
# == equal to comparision(operator)
a = 10
b = 10
result = (a==b)
print(result)
```

```python
str1 = "Roshan"
str2 = "roshan"

str1 == str2
```

```python
# Not equal to
str1 != str2
```

```python
# greater than >

a  = 45
b = 55

a > b
```

```python
## less than <
a<b
```

```python
# greater than or equal to
num1 = 55
num2 = 45

num1 >= num2
```

```python
# less than or equal to
num1 <= num2
```

### Logical Operators

```python
# And, OR, Not
# And(both condition should be true )
x = True
y = True

result = x and y
print(result)
```

```python
x = True
y = False

result = x and y
print(result)
```

```python
# OR 
x = True
y = True

result = x or y
print(result)
```

```python
x = True
y = False

result = x or y
print(result)
```

```python
# Not Operator( it is the inverse of the boolean variable)
x = True
not x
```

```python
# Simple calculator

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
modulus = num1 % num2
exponentiation = num1 ** num2

# displaying result
print(addition)
print(subtraction)
print(multiplication)
print(division)
print(floor_division)
print(modulus)
print(exponentiation)
```
