# Loops
In Python, loops are control flow statements used to repeat a block of code multiple times, enabling efficient execution of task over sequences or based on conditions. There are two primary types of loops:

i. for loop

ii. while loop

### For Loop
for loop is used used to iterate over sequence of number

```python
# 
range(5)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,10,2):  # (start, stop, step)
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
## string
str = "Roshan Bohara"
for i in str:
    print(i)
```

### while loop
The while loop continues to execute as long as the condition is True.

```python
### while loop
count = 0
while count < 5:
    print(count)
    count = count+1
```

```python
count = 0
while count%2 == 0:
    print(count)
    count = count + 1
```

### loop control statement
- break
- continue 
- pass

```python
## Break
# The break statement exits the loop permanently

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## Continue
# The continue statement skips the surrent iteration and continues with the next

for i in range(10):
    if i%2==0: # skips
        continue
    print(i)
```

```python
## pass
# THe pass statement is a null operation: it does nothing.

for i in range(5):
    if i == 3:
        pass
    print(i)
```

### Nested Loop
Nested loop is nothing but a loop inside a loop

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Calculate the sum of first N natural numbers using a while and for loop
# while loop
n = 10
sum = 0
count = 1
while count <=n:
    sum = sum+count
    count = count + 1

print("Sum of first n nuatural number:", sum)
```

```python
# for loop
n = 10
sum = 0
for i in range(11):
    sum = sum + i
print(sum)
```

```python
## Prime numbers between 1 and 100

for num in range(1,10):
    if num>1:
        for i in range(2, num):
            if num%i == 0:
                break
        else:
            print(num)
```

## Conclusion
Loops are powerful constructs in Python that allow you to execute a block of code multiple times. By understanding and using for and while loops, along with loop control statements like break, continue and pass, you can handle a wide range of programming tasks efficiently.
